```python
from diffusers import StableDiffusionPipeline
import torch
```

```python
def load_sd_pipeline():
  pipe = StableDiffusionPipeline.from_pretrained( #downloads and loads a pre-trained model called runwayml/stable-diffusion-v1-5.
      "runwayml/stable-diffusion-v1-5",
      torch_dytpe = torch.float16 #using 16-bit floating point precision
  ).to("gpu") # Moves the entire pipeline to the GPU for computation.

  return pipe # Returns the loaded pipeline
```

```python
def generate_image(prompt, output_File):
  pipe = load_sd_pipeline() # loads the Stable Diffusion model pipeline into memory.
  image = pipe(prompt).images[0] #pipe(prompt) runs the model and returns an object with a list of images. we select the very first image
  image.save(output_File) #saving the image to local directory
  print(f"Saved: {output_File}")
```

```python
if __name__ == "__main__":
    base_prompt = input("Enter the Keyword for image : ")

    print("Generating Image ... ")
    generate_image(base_prompt, f"{base_prompt}.png")
```
